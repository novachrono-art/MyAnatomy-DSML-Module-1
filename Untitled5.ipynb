```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("sudalairajkumar/covid19-in-india")

print("Path to dataset files:", path)
```

```python
import os
files = os.listdir(path)
print("Files in dataset:", files)
```

```python
import pandas as pd
file_path = os.path.join(path, "covid_19_india.csv")
df = pd.read_csv(file_path)
print(df.head())
```

```python
# understand dataset
print("shape", df.shape)
print("columns", df.columns)
print("datatype",df.dtypes)
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates()
```

```python
df["Date"] = pd.to_datetime(df["Date"])
df.info()
```

```python
df["Recovered"]=df["Cured"]
```

```python
#create a new column
df["Active"] =df["Confirmed"]-df["Recovered"]-df["Deaths"]
df.info()
df.tail()
```

```python
df.sample(100)
```

```python
state_data = (df.groupby("State/UnionTerritory")[["Confirmed","Deaths","Recovered","Active"]].max().reset_index())
state_data.head()
```

```python
#now calculate recovery rate
import numpy as np
state_data["Recovery_Rate"] = state_data["Recovered"]/state_data["Confirmed"].replace(0,np.nan)*100
```

```python
state_data["Death_Rate"] = state_data["Deaths"]/state_data["Confirmed"].replace(0,np.nan)*100
```

```python
state_data.head()
```

```python
state_data["Death_Rate"] = state_data["Death_Rate"].round(2)
state_data["Recovery_Rate"] = state_data["Recovery_Rate"].round(2)
state_data.head()
```

```python
# Total Cases by State
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(12,5))
sns.barplot(
    data = state_data.sort_values("Confirmed",ascending=False),
    x="State/UnionTerritory",
    y="Confirmed"
)
plt.xticks(rotation=90)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
# Top 10 states
top10_states = state_data.sort_values(by="Confirmed",ascending=False).head(10)
plt.figure(figsize=(12, 5))
sns.barplot(data=top10_states,x="State/UnionTerritory",y="Confirmed",palette=["Green","Pink","Blue"])
plt.title("Top 10 States")
# plt.xlabel("State/Union Territory")
# plt.ylabel("Confirmed Cases")
plt.xticks(rotation=90)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
# Top 10 states
top10_states = state_data.sort_values(by="Death_Rate",ascending=False).head(10)
plt.figure(figsize=(12, 5))
sns.barplot(data=top10_states,x="State/UnionTerritory",y="Death_Rate",palette=["Green","Pink","Blue"])
plt.title("Top 10 States")
plt.xticks(rotation=90)
plt.show()
```

```python
# Recovery Rate vs Death Rate
rate_data = state_data.sort_values(by="Recovery_Rate",ascending=False).head(10)
plt.figure(figsize=(10,6))
sns.barplot(data=rate_data,x="State/UnionTerritory",y="Recovery_Rate",palette=["Green","Pink","Blue"])
plt.title("Recovery Rate")
plt.xticks(rotation=90)
plt.show()

```
